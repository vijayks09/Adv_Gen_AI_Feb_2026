## TASK 1: Conceptual Understanding (Written)

>### 1. "Love" vs "love"
- NLP treats them as different tokens if case-sensitive
- Lowercasing makes them same → normalization
>- Example:
- "Love" ≠ "love" (before preprocessing)
- "love" = "love" (after lowercasing)

>### 2. If stopwords are NOT removed:
- Model gets noise
- Increased dimensionality
- Reduced performance

>### 3. When removing stopwords is harmful:
- Sentiment analysis → "not good" → removing "not" breaks meaning
- Chatbots → "how are you" → removing "are" affects grammar

>### 4. Stemming vs Lemmatization:
>- Stemming	- Lemmatization
>- Cuts words crudely	- Uses dictionary
>- "running" → "run" or "runn"	- "running" → "run"
>- Faster	- More accurate

## TASK 2: Build preprocess_text(text)

```python
import re

def preprocess_text(text):
    if not isinstance(text, str) or text.strip() == "":
        return [], ""
    
    # 1. Remove URLs & emails
    text = re.sub(r"http\S+|www\S+|@\S+", "", text)
    
    # 2. Remove numbers
    text = re.sub(r"\d+", "", text)
    
    # 3. Lowercase
    text = text.lower()
    
    # 4. Handle repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    
    # 5. Remove special characters (keep alphabets & spaces)
    text = re.sub(r'[^a-z\s]', '', text)
    
    # 6. Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    # 7. Tokenization
    tokens = text.split()
    
    # 8. Remove short tokens (≤2 except "no", "not")
    tokens = [word for word in tokens if len(word) > 2 or word in ["no", "not"]]
    
    # Final cleaned sentence
    clean_sentence = " ".join(tokens)
    
    return tokens, clean_sentence
```

## TASK 3: Stress Testing
>### Step 1: Prepare test data

```python
test_sentences = [
"Get 100% FREE access now!!!",
"I absolutely looooved this product 😍😍",
"Worst service ever... 0/10",
"Call me at 9876543210",
"This is THE best course!!!",
"Visit https://openai.com now!",
"Nooooo this is baaad!!!",
"OK OK OK I got it",
"Win $$$ now!!! Limited offer!!!",
"I am not happy with this"
]
```

>### Step 2: Run and print results

```python
for sentence in test_sentences:
    tokens, clean = preprocess_text(sentence)
    
    print("Original:", sentence)
    print("Tokens:", tokens)
    print("Clean:", clean)
    print("-"*50)
```

## TASK 4: Token Analytics

```python
def token_analysis(tokens):
    total = len(tokens)
    unique = len(set(tokens))
    avg_len = sum(len(word) for word in tokens) / total if total > 0 else 0
    
    return total, unique, avg_len

# Apply
for sentence in test_sentences:
    tokens, _ = preprocess_text(sentence)
    
    total, unique, avg = token_analysis(tokens)
    
    print(sentence)
    print("Total:", total, "Unique:", unique, "Avg Length:", avg)
    print("-"*40)
```

>- Answer analysis questions:
- Most noise: sentences with emojis, repeated chars, URLs
- Most meaningful tokens: clean sentences with actual words (e.g., sentiment sentences)

## TASK 5: Frequency Analysis

```python
from collections import Counter

all_tokens = []

for sentence in test_sentences:
    tokens, _ = preprocess_text(sentence)
    all_tokens.extend(tokens)

counter = Counter(all_tokens)

print("Top 10 words:", counter.most_common(10))
print("Least 5 words:", counter.most_common()[-5:])
```

## TASK 6: Full Pipeline

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []
    
    for text in text_list:
        tokens, clean = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean)
    
    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```

```python
output = full_pipeline(test_sentences)
print(output)
```
